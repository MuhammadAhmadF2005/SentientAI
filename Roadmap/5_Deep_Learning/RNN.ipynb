# Recurrent Neural Networks (RNN)
A simple and intuitive guide to understanding and implementing Simple/Vanilla Recurrent Neural Networks (RNNs).

### Core Concept
Unlike feedforward neural networks, RNNs process sequential data by keeping a **hidden state** ($h_t$) that acts as memory across time steps:
$$h_t = \tanh(W_{hh} h_{t-1} + W_{xh} x_t + b_h)$$
$$y_t = W_{hy} h_t + b_y$$

## Part 1: Manual Vanilla RNN Forward Pass (NumPy)
Step-by-step implementation of a single RNN step from scratch using NumPy to understand the math.

```python
import numpy as np

# Hyperparameters
input_size = 3   # Feature dimension per timestep
hidden_size = 4  # Number of hidden units
time_steps = 5   # Length of input sequence

# Random input sequence: shape (time_steps, input_size)
np.random.seed(42)
X = np.random.randn(time_steps, input_size)

# Initialize weights and bias
W_xh = np.random.randn(hidden_size, input_size)
W_hh = np.random.randn(hidden_size, hidden_size)
b_h  = np.zeros((hidden_size, 1))

# Initial hidden state
h_t = np.zeros((hidden_size, 1))

print("--- Manual Forward Pass through Time Steps ---")
for t in range(time_steps):
    x_t = X[t:t+1, :].T  # Shape: (input_size, 1)
    # Recurrent formula: h_t = tanh(W_hh * h_{t-1} + W_xh * x_t + b_h)
    h_t = np.tanh(np.dot(W_hh, h_t) + np.dot(W_xh, x_t) + b_h)
    print(f"Timestep {t+1} Hidden State:", np.round(h_t.flatten(), 3))
```

## Part 2: Simple RNN with TensorFlow / Keras
Building and training a SimpleRNN model to predict the next number in a sequence.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# Generate sequence data: e.g. [0, 1, 2] -> 3, [1, 2, 3] -> 4
X_train = np.array([[[i+j] for j in range(3)] for i in range(100)], dtype=np.float32)
y_train = np.array([[i+3] for i in range(100)], dtype=np.float32)

# Define SimpleRNN Model
model = Sequential([
    SimpleRNN(units=16, activation='tanh', input_shape=(3, 1)),
    Dense(units=1)
])

model.compile(optimizer='adam', loss='mse')
model.summary()

# Train model
history = model.fit(X_train, y_train, epochs=30, verbose=0)
print("Training Complete. Final Loss:", round(history.history['loss'][-1], 4))

# Test Prediction
test_input = np.array([[[10], [11], [12]]], dtype=np.float32)
predicted = model.predict(test_input, verbose=0)
print(f"Input: [10, 11, 12] -> Predicted Next Value: {predicted[0][0]:.2f} (Expected: 13.0)")
```
